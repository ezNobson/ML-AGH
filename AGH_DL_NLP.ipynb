# 1) Understanding text

```python
texts = [
    "Odkryto złoża żelaza w warstwach osadowych",
    "Badania geofizyczne wskazują anomalię magnetyczną",
    "Misja kosmiczna potwierdziła obecność minerałów na Marsie",
    "Wiercenia w rejonie rud miedzi wykazały wysoką koncentrację metalu",
    "Sonda pobrała próbki gruntu z księżyca Jowisza",
    "Analiza sejsmiczna wskazuje na pęknięcia w skorupie ziemskiej"
]
print("Number of sentences", len(texts))
```

Most important words:
1. żelazno, osadowych
2. badania, anomalia magnetyczna
3. misja kosmiczna, mars
4. rud miedzi, koncentracja metalu
5. sonda, księżyc Jowisza
6. Analiza sejsmiczna, skorupa ziemska

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(num_words=100)  # ograniczamy słownik do 100 najczęściej występujących słów
tokenizer.fit_on_texts(texts)          # wstaw listę tekstów
print(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(texts)  # wstaw listę tekstów
print(sequences)
```

```python
first_seq = sequences[0]
print(f"\nPierwsze zdanie: {texts[0]}")
print(f"Liczby: {first_seq}")
```

```python
reverse_word_index = {v: k for k, v in tokenizer.word_index.items()}
decoded_words = [reverse_word_index.get(i) for i in first_seq]
print(f"Odkodowane: {decoded_words}")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded = pad_sequences(sequences, padding='post')  # dopasowanie długości poprzez dodanie zer na końcu
print(padded)
```

```python
texts = [
    "Odkryto złoża żelaza w warstwach osadowych regionu X",
    "Wiercenia w rejonie rud miedzi wykazały wysoką koncentrację metalu",
    "Misja kosmiczna potwierdziła obecność minerałów na Marsie",
    "Badania geofizyczne wykazały anomalię magnetyczną",
    "Sonda pobrała próbki gruntu z księżyca Jowisza",
    "Analiza sejsmiczna wskazuje na pęknięcia w skorupie ziemskiej"
]
labels = np.array([0, 0, 1, 1, 1, 0])  # 0 = geologia / złoża, 1 = kosmos / misje
```

```python
tokenizer = Tokenizer(num_words=100)
tokenizer.fit_on_texts(texts)
sequences_train = tokenizer.texts_to_sequences(texts)
```

```python
padded_train = pad_sequences(sequences_train, padding='post')  # dopasowanie długości poprzez dodanie zer na końcu
print(padded)
```

# 2) Simple model in Keras

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Flatten, Dense
```

```python
model = Sequential([
    Embedding(input_dim=100, output_dim=8, input_length=len(padded_train[0])),  # Embedding layers is to transform integer number for example 5 into list of loat nuimbers 5 -> [0.1, -0.05, 0.8, ...]
    Flatten(),
    Dense(1, activation='sigmoid')  # dla klasyfikacji binarnej
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(padded_train, labels, epochs=10)
```

```python
test_text = ["Nowe badania wskazują złoża niklu na asteroidzie"]
seq = tokenizer.texts_to_sequences(test_text)
padded_test = pad_sequences(seq, maxlen=len(padded_train[0]), padding='post')
```

```python
prediction = model.predict(padded_test)
prediction
```

```python
print(f"Prediction results: {prediction[0][0]:.4f}")
```

```python
if prediction < 0.5:
    print("Geology")
else:
    print("Cosmos")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
```

```python
def tokenize_pad(X):
    seq = tokenizer.texts_to_sequences(X)
    return pad_sequences(seq, maxlen=len(padded_train[0]), padding='post')
```

```python
pipeline = Pipeline([
    ('tokenize_pad', FunctionTransformer(tokenize_pad, validate=False)),
    ('model', model)
])
```

```python
new_texts = ["Eksploracja Marsa wykazała obecność żelaza i niklu"]
```

```python
prediction = pipeline.predict(new_texts)
prediction
```

```python
if prediction[0][0] > 0.5:
  print("Cosmos")
else:
  print("Geology")
```

# 3) Addidtional cleaning function

```python
def clean_fun(text_list):
  cleaned =[]
  stop_words = ["w", "i", "na", "z"]
  for text in text_list:
    text = text.lower()
    for stop_word in stop_words:
      text = text.replace(stop_word, "")
    cleaned.append(text)
  return cleaned
```

```python
pipeline_stop_words = Pipeline([
    ('clean', FunctionTransformer(clean_fun, validate=False)),
    ('tokenize_pad', FunctionTransformer(tokenize_pad, validate=False)),
    ('model', model)
])
```
